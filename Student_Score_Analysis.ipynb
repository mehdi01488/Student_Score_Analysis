```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("student_scores.csv")
print(data.head())
```

```python
data.isnull().sum()
```

##### Drop unnamed column

```python
data = data.drop("Unnamed: 0", axis = 1)
data
```

#### gender distribution

```python
plt.figure(figsize = (5,5))
ax = sns.countplot(data = data, x = "Gender", color = "cyan")
ax.bar_label(ax.containers[0] )
plt.title("Gender Distribution")
plt.show()
```

# from the above chart we have analysed that 
# the number of females in the data is  more than the number of male

```python
gb = data.groupby("ParentEduc").agg({"MathScore": "mean" ,	"ReadingScore": "mean"	,"WritingScore": "mean"})
gb
```

```python
plt.figure(figsize = (4,4))
sns.heatmap(gb , annot = True)
plt.title("Relationship between Parent's Education and Student's Score")
plt.show()
```

## from the above chart we have concluded that the education
## of the parents have a good impact on their Scores

```python
gb1 = data.groupby("ParentMaritalStatus").agg({"MathScore": "mean" ,	"ReadingScore": "mean"	,"WritingScore": "mean"})
gb1
```

```python
plt.figure(figsize = (4,4))
sns.heatmap(gb1 , annot = True)
plt.title("Relationship between ParentMaritalStatus and Student's Score")
plt.show()
```

## from the above chart we have analysis the MaritalStatus
## have no impact or negligible impact on the student scores

### Distribuation of EthnicGroup

```python
groupA = data.loc[(data["EthnicGroup"] == "group A")].count()
groupB = data.loc[(data["EthnicGroup"] == "group B")].count()
groupC = data.loc[(data["EthnicGroup"] == "group C")].count()
groupD = data.loc[(data["EthnicGroup"] == "group D")].count()
groupE = data.loc[(data["EthnicGroup"] == "group E")].count()

l = ["group A" , "group B", "group C", "group D", "group E"]
mlist = [groupA["EthnicGroup"] , groupB["EthnicGroup"],groupC["EthnicGroup"],groupD["EthnicGroup"], groupE["EthnicGroup"]]

plt.pie(mlist, labels = l , autopct = ("% 1.2f%%") , shadow = True , explode =  [0.0 , 0. , 0.1 ,0.0 , 0.0] )
plt.title("Distribuation of EthnicGroup")
plt.show()
```
